```python
from dataclasses import dataclass
from datetime import datetime
from typing import Optional

@dataclass
class Article:
    source: str
    title: str
    url: str
    published_at: Optional[datetime]
    content: Optional[str]
    language: str
    summary: Optional[str] = None
```

```python
import requests
import feedparser
import urllib.parse
from datetime import datetime
import os

BBC_API_BASE = "http://localhost:3000/api"

def fetch_bbc_bengali(limit=10):
    try:
        resp = requests.get(f"{BBC_API_BASE}/news", timeout=10)
        resp.raise_for_status()
        data = resp.json()
    except Exception as e:
        print(f"[bbc_bengali] fetch failed: {e}")
        return []
    articles = []
    for item in data.get("articles", [])[:limit]:
        articles.append(Article(
            source="bbc_bengali", title=item.get("title", ""), url=item.get("link", ""),
            published_at=None, content=item.get("description"), language="bn",
        ))
    return articles


ABP_FEEDS = {
    "home": "https://bengali.abplive.com/home/feed",
    "kolkata": "https://bengali.abplive.com/news/kolkata/feed",
    "states": "https://bengali.abplive.com/states/feed",
    "district": "https://bengali.abplive.com/district/feed",
}

def fetch_abp_rss(section="kolkata", limit=10):
    url = ABP_FEEDS.get(section)
    if not url:
        print(f"[abp] unknown section: {section}")
        return []
    feed = feedparser.parse(url)
    articles = []
    for entry in feed.entries[:limit]:
        published = datetime(*entry.published_parsed[:6]) if hasattr(entry, "published_parsed") and entry.published_parsed else None
        articles.append(Article(
            source="abp_ananda", title=entry.get("title", ""), url=entry.get("link", ""),
            published_at=published, content=entry.get("summary"), language="bn",
        ))
    return articles


def fetch_google_news_rss(query, lang="bn-IN", limit=10):
    gl = lang.split("-")[-1]
    ceid = f"{gl}:{lang.split('-')[0]}"
    encoded_query = urllib.parse.quote(query)
    url = f"https://news.google.com/rss/search?q={encoded_query}&hl={lang}&gl={gl}&ceid={ceid}"
    feed = feedparser.parse(url)
    articles = []
    for entry in feed.entries[:limit]:
        published = datetime(*entry.published_parsed[:6]) if hasattr(entry, "published_parsed") and entry.published_parsed else None
        articles.append(Article(
            source="google_news", title=entry.get("title", ""), url=entry.get("link", ""),
            published_at=published, content=entry.get("summary"), language=lang.split("-")[0],
        ))
    return articles


from tavily import TavilyClient

def fetch_tavily(query, days=1, limit=10, include_reddit=False):
    try:
        client = TavilyClient(api_key=os.environ["TAVILY_API_KEY"])
        kwargs = {"query": query, "search_depth": "advanced", "max_results": limit, "days": days, "include_raw_content": True}
        if include_reddit:
            kwargs["include_domains"] = ["reddit.com"]
        response = client.search(**kwargs)
    except Exception as e:
        print(f"[tavily] fetch failed: {e}")
        return []
    articles = []
    for r in response.get("results", []):
        articles.append(Article(
            source="tavily_reddit" if include_reddit else "tavily_news",
            title=r.get("title", ""), url=r.get("url", ""), published_at=None,
            content=r.get("content") or r.get("raw_content"), language="en",
        ))
    return articles


def plan_and_search(topic="West Bengal"):
    queries = [f"{topic} latest news", f"{topic} Kolkata"]
    results = []
    for q in queries:
        results += fetch_tavily(q, days=1, limit=8)
    if len(results) < 5:
        results += fetch_tavily(f"{topic}", days=3, limit=10)
    return results
```

```python
import google.generativeai as genai
import numpy as np
from itertools import combinations

genai.configure(api_key=os.environ["GEMINI_API_KEY"])
SIMILARITY_THRESHOLD = 0.85  # Gemini embeddings tend to cluster tighter — retune if needed

def get_embeddings(texts):
    result = genai.embed_content(model="models/gemini-embedding-001", content=texts)
    return np.array(result["embedding"])

def cosine_sim(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

def dedup_articles(articles):
    if len(articles) <= 1:
        return articles
    texts = [f"{a.title}. {(a.content or '')[:200]}" for a in articles]
    try:
        embeddings = get_embeddings(texts)
    except Exception as e:
        print(f"[dedup] embedding call failed, skipping dedup: {e}")
        return articles

    n = len(articles)
    parent = list(range(n))
    def find(x):
        while parent[x] != x:
            parent[x] = parent[parent[x]]
            x = parent[x]
        return x
    def union(x, y):
        parent[find(x)] = find(y)
    for i, j in combinations(range(n), 2):
        if cosine_sim(embeddings[i], embeddings[j]) > SIMILARITY_THRESHOLD:
            union(i, j)
    clusters = {}
    for i in range(n):
        clusters.setdefault(find(i), []).append(i)
    deduped = [articles[max(idxs, key=lambda idx: len(articles[idx].content or ""))] for idxs in clusters.values()]
    print(f"[dedup] {n} articles -> {len(deduped)} after clustering")
    return deduped
```

```python
import re
import json
from groq import Groq

groq_client = Groq(api_key=os.environ["GROQ_API_KEY"])
GROQ_MODEL = "llama-3.3-70b-versatile"

def build_relevance_batch_prompt(articles, topic="West Bengal"):
    lines = [f"{i}: {a.title} — {(a.content or '')[:150].strip()}" for i, a in enumerate(articles)]
    joined = "\n".join(lines)
    return f"""Below is a numbered list of article titles and snippets. For each, decide if it is
genuinely about {topic} specifically (current politics, people, places, culture, or events) —
NOT a generic wiki/tourism/reference page, NOT primarily about Bangladesh, NOT a homepage/section listing.

{joined}

Respond with ONLY a JSON array of the relevant numbers, nothing else. Example: [0, 2, 5, 7]"""

def filter_relevant(articles, topic="West Bengal"):
    if not articles:
        return []
    prompt = build_relevance_batch_prompt(articles, topic)
    try:
        completion = groq_client.chat.completions.create(
            model=GROQ_MODEL,
            messages=[{"role": "user", "content": prompt}],
            max_tokens=300,
            temperature=0.1,
        )
        raw = completion.choices[0].message.content.strip()
        match = re.search(r"\[[\d,\s]*\]", raw)
        relevant_indices = set(json.loads(match.group())) if match else set(range(len(articles)))
    except Exception as e:
        print(f"[relevance] batch check failed, keeping all: {e}")
        return articles
    relevant = [a for i, a in enumerate(articles) if i in relevant_indices]
    print(f"[relevance] {len(articles)} -> {len(relevant)} relevant")
    return relevant
```

```python
def build_compose_prompt(articles, topic="West Bengal"):
    blocks = []
    for i, a in enumerate(articles, 1):
        blocks.append(f"{i}. [{a.source}] {a.title}\nContent: {(a.content or a.title)[:400]}\nURL: {a.url}")
    joined = "\n\n".join(blocks)
    return f"""You are composing a daily news digest about {topic} for a single reader.
Below are today's new, deduplicated, relevant articles with their raw content.

{joined}

Instructions:
- Group related articles under short category headers (e.g. Politics, Kolkata City, Sports, Culture, Weather).
- Order groups by importance, most significant news first.
- Write 1-2 original sentences summarizing each story yourself from the content given — don't just repeat it verbatim.
- Merge articles covering the same event into one entry rather than listing them separately.
- Keep it concise and skimmable — read on a phone.
- Include the source URL after each story.
- Do not invent facts not present in the content above.

Write the final digest now, in Markdown with headers."""

def _fallback_flat_digest(articles):
    lines = [f"Bengal News Digest — {len(articles)} new stories\n"]
    for a in articles:
        lines.append(f"\n[{a.source}] {a.title}\n{(a.content or '')[:200]}\n{a.url}")
    return "\n".join(lines)

def compose_digest(articles, topic="West Bengal"):
    if not articles:
        return "No new relevant stories today."
    prompt = build_compose_prompt(articles, topic)
    try:
        completion = groq_client.chat.completions.create(
            model=GROQ_MODEL,
            messages=[{"role": "user", "content": prompt}],
            max_tokens=1500,
            temperature=0.3,
        )
        return completion.choices[0].message.content.strip()
    except Exception as e:
        print(f"[compose_digest] failed, falling back: {e}")
        return _fallback_flat_digest(articles)
```

```python
import requests
import os
import re
import html

def clean_text(text):
    if not text:
        return ""
    text = html.unescape(text)              # &lt;p&gt; -> <p>
    text = re.sub(r"<[^>]+>", "", text)      # strip actual HTML tags
    return text.strip()

def send_telegram_digest(text):
    url = f"https://api.telegram.org/bot{os.environ['TELEGRAM_BOT_TOKEN']}/sendMessage"
    # split on double-newlines (article boundaries) instead of blind character slicing
    blocks = text.split("\n\n")
    chunks, current = [], ""
    for block in blocks:
        if len(current) + len(block) + 2 > 3500:
            chunks.append(current)
            current = block
        else:
            current += ("\n\n" if current else "") + block
    if current:
        chunks.append(current)

    for chunk in chunks:
        resp = requests.post(url, json={
            "chat_id": os.environ["TELEGRAM_CHAT_ID"],
            "text": chunk,
            "parse_mode": "Markdown"
        })
        if not resp.ok:
            print(f"[telegram] send failed: {resp.status_code} {resp.text}")
            # retry once as plain text if Markdown parsing was the issue
            resp2 = requests.post(url, json={
                "chat_id": os.environ["TELEGRAM_CHAT_ID"],
                "text": chunk,
            })
            resp2.raise_for_status()
    print("Digest sent to Telegram.")
```

```python
import sqlite3
import hashlib

conn = sqlite3.connect("digest.db")
conn.execute("""CREATE TABLE IF NOT EXISTS seen_articles (
    url_hash TEXT PRIMARY KEY, source TEXT, title TEXT, seen_at TEXT DEFAULT CURRENT_TIMESTAMP
)""")
conn.commit()

def url_hash(url):
    return hashlib.sha256(url.encode()).hexdigest()

def filter_new_articles(articles):
    new = []
    for a in articles:
        h = url_hash(a.url)
        if conn.execute("SELECT 1 FROM seen_articles WHERE url_hash = ?", (h,)).fetchone() is None:
            new.append(a)
    return new

def mark_seen(articles):
    for a in articles:
        conn.execute("INSERT OR IGNORE INTO seen_articles (url_hash, source, title) VALUES (?, ?, ?)",
                      (url_hash(a.url), a.source, a.title))
    conn.commit()
```

```python
def run_pipeline(topic="West Bengal"):
    print("Step 1: Ingesting from all sources...")
    articles = []
    articles += fetch_bbc_bengali(limit=10)
    articles += fetch_abp_rss(section="kolkata", limit=10)
    articles += fetch_google_news_rss(query=topic, lang="bn-IN", limit=10)
    articles += plan_and_search(topic=topic)
    print(f"Total raw articles: {len(articles)}")

    print("\nStep 2: Deduplicating...")
    articles = dedup_articles(articles)

    print("\nStep 3: Filtering relevance...")
    articles = filter_relevant(articles, topic=topic)

    print("\nStep 4: Checking against seen articles...")
    new_articles = filter_new_articles(articles)
    print(f"New articles: {len(new_articles)}")

    if not new_articles:
        print("No new articles — nothing to send.")
        return

    print("\nStep 5: Composing digest (summarizes + groups in one call)...")
    digest = compose_digest(new_articles, topic=topic)
    print("\n--- DIGEST PREVIEW ---\n")
    print(digest)

    print("\nStep 6: Sending to Telegram...")
    send_telegram_digest(digest)

    mark_seen(new_articles)
    print("\nDone. Marked as seen.")

run_pipeline(topic="West Bengal")
```
